```python
print("Hello World")

# Strings
name = "Alex"
# Integers (Whole Numbers)
age = 25
# Floats (Decimal Numbers)
height = 9.5

# Booleans (True/False)
is_student = False

# There are two ways to print variables in Python
print("Hey, my name is " + name)
print("Hey, my name is", name)

print(name[-1])
```

```python

# Remember that Python is a case-sensitive language, so "Hello" and "hello" are different variables/words
message = "hello world"
print(message.upper())  # This will convert the string to uppercase
print(message.lower())  # This will convert the string to lowercase
# This will convert the first letter of the string to uppercase
print(message.capitalize())
# This will replace all the l's in the string with L's
print(message.replace("l", "L"))
print("World" in message)  # This will check if "World" is in the string
print(len(message))  # This will print the length of the string

greeting1 = "Hello"
greeting2 = "hello"
```

```python
# Remember that in Python the identation is very important, so be careful with the indentation of your code
if greeting1 == greeting2:
    print("The greetings are the same")
else:
    print("The greetings are different")

# Type Conversion
age_str = "30"
age_num = int(age_str)
print(type(age_num))
print(type(age_str))

price_float = 29.99
price_int = int(price_float)
```
