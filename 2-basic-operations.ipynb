```python
x = 10
y = 5

print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x % y)  # Modulus operator (remainder of the division)
print(x ** y)  # Exponentiation operator (x raised to the power of y)
# Floor division operator (quotient of the division, rounded down to the nearest integer)
print(x // y)

# x = x + 15
x += 15  # This is the same as x = x + 15
print(x)
```

```python
# Concatenation of strings
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name  # Concatenation of strings
print(full_name)


print("Hey, my name is " + full_name)  # Concatenation of strings
print("Hey, my name is", full_name)  # Concatenation of strings
```

```python
# f-strings
# f-strings are a way to format strings in Python, and they are very useful for concatenating strings
print(f"Hey, my name is {full_name}")


# int floor division (essenstially floor division rounds down to the nearest integer)
a = 17
b = 5

print(a/b)  # This will print 3.4
print(a//b)  # This will print 3, because it is the floor division (the quotient of the division, rounded down to the nearest integer)

# This is useful when you want to assign a value to multiple variables at once
i, j, k = 1, 2, 3
print(i, j, k)  # This will print 1 2 3
```

```python
# Swap values
m = 10
n = 20
m, n = n, m
print(m, n)  # This will print 20 10
```

```python
# Comparison Operators
c = 5
d = 10
print(c == d)  # This will print False, because c is not equal to d
print(c != d)  # This will print True, because c is not equal to d
print(c < d)  # This will print True, because c is less than d
print(c > d)  # This will print False, because c is not greater than d
print(c <= d)  # This will print True, because c is less than or equal to d
print(c >= d)  # This will print False, because c is not greater than or equal to d
```

```python
# Logical Operators
# Logical operators are used to combine conditional statements

a = True
b = False

print(a and b)  # This will print False, because both a and b are not True
print(a or b)  # This will print True, because at least one of a or b is True
print(not a)  # This will print False, because a is True and not True is False
print(not b)  # This will print True, because b is False and not False is True
```
